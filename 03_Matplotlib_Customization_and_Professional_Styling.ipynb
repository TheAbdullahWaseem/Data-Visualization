##3.2 figsize
## What is figsize?
figsize controls the width and height of the plot (in inches).

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 4))

plt.plot([1, 2, 3, 4], [10, 15, 13, 20])

plt.title("Plot with Figure Size")
plt.show()
```

- Large figures → clear and professional

##3.3 DPI (Image Quality)
## What is DPI?
DPI = Dots Per Inch

It controls the sharpness of the image.

```python
plt.figure(figsize=(6,4), dpi=100)

plt.plot([1, 2, 3, 4], [10, 15, 13, 20])

plt.title("High Quality Plot")
plt.show()
```

- Low DPI → blurry image
- High DPI → sharp, print-ready image

###Professional Rule
- Notebook work → dpi 80–100
- Reports / export → dpi 150–300

##3.4 Tight Layout
## What is tight_layout?
It automatically adjusts spacing between plot elements.
### Problem without tight_layout
- Labels overlap
- Titles get cut
- Axis labels look cramped

```python
plt.figure(figsize=(6,4))

plt.plot([1, 2, 3, 4], [10, 15, 13, 20])

plt.title("Without Tight Layout")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.tight_layout()

plt.show()
```

- Fixes spacing automatically
- Makes plots look clean in reports
- Prevents label overlapping

##3.5 Combined Professional Example

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5), dpi=120)

plt.plot([1, 2, 3, 4], [10, 15, 13, 20],
         color='blue',
         marker='o',
         linestyle='--',
         linewidth=2)

plt.title("Professional Plot Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.grid(True)

plt.tight_layout()

plt.show()
```

###Key Takeaways

#6. Legends
## 6.1 What is a Legend?
A legend is a label box that explains what each line or data series represents in a plot.

 In simple terms:

Legend answers: “Which line is which?”

###Example (Without Legend Problem)

```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3], [10, 20, 15])
plt.plot([1, 2, 3], [5, 15, 10])

plt.title("Without Legend")
plt.show()
```

##Correct Version (With Labels + Legend)

```python
plt.plot([1, 2, 3], [10, 20, 15], label="Sales")
plt.plot([1, 2, 3], [5, 15, 10], label="Profit")

plt.title("With Legend Example")
plt.legend()

plt.show()
```

###Key Insight
Without legend → data is unclear  
With legend → data becomes interpretable

##6.2 Proper Legend Placement
### Why placement matters
If legend covers data → chart becomes hard to read

If placed correctly → improves clarity

###Example Changing Location

```python
plt.plot([1, 2, 3], [10, 20, 15], label="Sales")
plt.plot([1, 2, 3], [5, 15, 10], label="Profit")

plt.legend(loc="upper left")

plt.show()
```

###Common locations
- "best" → automatic (recommended)
- "upper right"
- "upper left"
- "lower right"
- "lower left"

###Professional Rule
Always use loc="best" unless it overlaps data.

Clutter = too much visual information that confuses the viewer

###Common mistakes
- Too many lines in one plot
- Long legend text

- Too many colors

###Example of clean legend

```python
plt.plot([1, 2, 3], [10, 20, 15], label="Sales")
plt.plot([1, 2, 3], [5, 15, 10], label="Profit")

plt.legend(loc="best", fontsize=9)

plt.show()
```

#7. Annotations (Insight Highlighting)
## 7.1 What is Annotation?
Annotation means adding text or markers directly on a plot to explain or highlight important insights.

 In simple words:

You are “talking inside the graph”.

###Why annotations are important
In real data analysis:
- Managers don’t read raw numbers
- They look for key insights quickly

Annotations help answer:
 "What is important here?"

##7.2 Adding Basic Text on Graph

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4]
y = [10, 15, 13, 20]

plt.plot(x, y, marker='o')

plt.title("Sales Trend")

plt.annotate("Growth starts here",
             xy=(2, 15),
             xytext=(1, 18))

plt.show()
```

###What is happening?
- xy → point you want to highlight
- xytext → where text is placed

##7.3 Highlighting Maximum Value

```python
x = [1, 2, 3, 4]
y = [10, 15, 13, 20]

plt.plot(x, y, marker='o')

max_index = y.index(max(y))

plt.annotate("Peak Sales",
             xy=(x[max_index], y[max_index]),
             xytext=(x[max_index], y[max_index] + 2),
             arrowprops=dict(facecolor='black'))

plt.title("Sales with Peak Highlight")
plt.show()
```

This tells the viewer:

 "This is the most important point in the dataset"

##7.4 Highlighting Minimum Value

```python
x = [1, 2, 3, 4]
y = [10, 15, 13, 20]

plt.plot(x, y, marker='o')

min_index = y.index(min(y))

plt.annotate("Lowest Point",
             xy=(x[min_index], y[min_index]),
             xytext=(x[min_index], y[min_index] - 3),
             arrowprops=dict(facecolor='red'))

plt.title("Sales with Low Point Highlight")
plt.show()
```

##7.5 Drawing Attention to Insights

You don’t annotate everything.

You highlight only:
- Peaks
- Drops
- Anomalies
- Important changes

```python
plt.plot([1, 2, 3, 4], [10, 15, 13, 20], marker='o')

plt.title("Monthly Sales Trend")

plt.annotate("Drop in performance",
             xy=(3, 13),
             xytext=(2.5, 16),
             arrowprops=dict(facecolor='red'))

plt.show()
```

###Professional Insight
Annotation is NOT decoration.

It is:

explanation of data behavior

##7.6 Common Mistakes
- Annotating every point (clutter)
- Using unclear text
- Poor positioning
- No business meaning

##7.7 Key Takeaways
- Annotation adds meaning to data
- Highlight only important points
- Use arrows for clarity
- Focus on business insights, not decoration

#9. Subplots (Basic Introduction)
## 9.1 What are Subplots?
Subplots allow you to create multiple charts within a single figure (canvas).

Instead of creating separate graphs, you can organize multiple visualizations together.

##Why subplots are important
In real-world analysis, you often need to:

- Compare multiple variables
- Show different perspectives of data
- Present insights in one view

Subplots make this possible.

##9.2 Basic Structure of Subplots
###Example (2 plots in one figure)

```python
plt.style.use("default")
fig, ax = plt.subplots(1, 2, figsize = (10, 4), dpi = 200)

ax[0].plot([1,2,3],[10,20,15])
ax[0].set_title('Sales')

ax[1].bar([1,2,3],[5,7,6])
ax[1].set_title("Profit")

plt.tight_layout()
plt.show()
```

###What is happening?
- fig → overall canvas
- axes → individual plots
- axes[0], axes[1] → separate charts

##9.3 Layout Understanding
### Grid system
plt.subplots(rows, columns)

Example:

(1, 2) → 1 row, 2 plots

(2, 2) → 4 plots (2x2 grid)

###Example (2x2 layout)

```python
fig, axes = plt.subplots(2, 2, figsize=(10,8))

axes[0,0].plot([1,2,3], [10,15,20])
axes[0,0].set_title("Plot 1")

axes[0,1].bar([1,2,3], [5,7,6])
axes[0,1].set_title("Plot 2")

axes[1,0].hist([1,2,2,3,3,3])
axes[1,0].set_title("Plot 3")

axes[1,1].scatter([1,2,3], [4,5,6])
axes[1,1].set_title("Plot 4")

plt.tight_layout()
plt.show()
```

##9.4 When to Use Subplots
Use subplots when:

- Comparing multiple metrics (Sales vs Profit vs Cost)
- Showing different types of analysis
- Building dashboards in reports
- Presenting multiple insights together

###Real-world example
A business report may include:

- Sales trend (line plot)
- Product comparison (bar chart)
- Distribution (histogram)

 All in ONE figure using subplots

##9.6 Key Takeaways
- Subplots = multiple charts in one figure
- Use grid system (rows, columns)
- Always maintain readability
- Ideal for comparison and reporting

#10.2 savefig() (Core Function)
## What is savefig?
savefig() is used to save a plot as an image file.

```python
import matplotlib.pyplot as plt

plt.plot([1, 2, 3], [10, 20, 15])

plt.title("Sales Trend")

plt.savefig("sales_plot.png")

plt.show()
```

###Important Rule
Always call savefig() BEFORE plt.show()

##10.3 High-Quality Export
 Example with quality control

```python
plt.figure(figsize=(8,4), dpi=150)

plt.plot([1, 2, 3], [10, 20, 15], marker='o')

plt.title("Professional Plot")

plt.savefig("high_quality_plot.png", dpi=300)

plt.show()
```

###What’s happening?
- figsize → controls size
- dpi (figure) → screen quality
- dpi (savefig) → export quality (VERY IMPORTANT)

##10.4 PNG vs JPG
### PNG (Recommended)
- High quality
- Lossless (no data loss)
- Best for:
  → reports
  → dashboards
  → portfolio

###JPG
- Smaller file size
- Slight quality loss
- Best for:
  → quick sharing
  → non-critical visuals

###Professional Rule
Always use PNG for:

- portfolio

##10.5 DPI
## What is DPI again?
DPI (Dots Per Inch) controls image sharpness.

###Quality levels
- 80–100 → low quality (not professional)

- 300 → high quality (recommended)

###Professional Insight
A blurry plot = unprofessional work

##10.6 Complete Professional Example

```python
plt.figure(figsize=(10,5), dpi=120)

plt.plot([1, 2, 3, 4], [10, 15, 13, 20],
         marker='o',
         linestyle='--',
         color='blue',
         label='Sales')

plt.title("Monthly Sales Trend", fontsize=14, pad=15)
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()
plt.grid(True)

plt.tight_layout()

plt.savefig("sales_analysis.png", dpi=300)

plt.show()
```

##10.7 Key Takeaways

- Always use high DPI (300 for portfolio)
