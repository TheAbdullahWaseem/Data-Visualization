## 2. Core Concept: Figure vs Axes

When you use:

```python
fig, axes = plt.subplots()

###You are creating two layers:

- Figure (fig) → the entire canvas (dashboard container)
- Axes (axes) → individual plots inside the canvas
- Figure → complete report page
- Axes → individual charts inside that report

##3. Case 1: 1D Subplots (1 Row Layout)
###When do we use this?

We use a 1D layout when we want simple comparisons, such as:

- Product A vs Product B
- Before vs After analysis
- Two strategies comparison

###Structure: 1x2 Layout

###Plotting on 1D Axes


````python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(10, 4))

months = [1, 2, 3, 4, 5]
product_A = [10, 20, 25, 30, 40]
product_B = [12, 18, 22, 35, 38]

axes[0].plot(months, product_A, label = 'Product_A', marker = 'o')
axes[0].set_title("Product A Sales Trend")
axes[0].set_xlabel("Months")
axes[0].set_ylabel("Sales")
axes[0].grid(True, linestyle = '--', alpha = 0.6)

axes[1].plot(months, product_B, marker='o', color='orange')
axes[1].set_title("Product B Sales Trend")
axes[1].set_xlabel("Months")
axes[1].set_ylabel("Sales")
axes[1].grid(True)

plt.tight_layout()
plt.show()
````

###Key Insight: 1D Axes Behavior

###When you use:

``plt.subplots(1, 2)``

The axes object becomes a 1D array.

So you access plots like:

- axes[0]
- axes[1]

##Analyst Interpretation

This layout is best used for:

- Direct comparisons
- Performance comparison between two entities
- A/B testing visualization

##4. Case 2: 2D Subplots (Grid Layout)
###When do we use this?

We use a 2D layout when we want to build a complete analytical dashboard, including multiple perspectives:

- Trend analysis
- Comparison
- Distribution
- Relationship

 Structure: 2x2 Dashboard

````python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

x = np.linspace(0, 10, 100)
sales = np.random.randint(20, 100, 100)

#Top-Left: Trend Analysis

axes[0, 0].plot(x, np.sin(x), label = "Trend")
axes[0, 0].set_title("Trend Analysis")
axes[0, 0].grid(True)
axes[0, 0].legend()


#Top-Right: Category Comparison

categories = ["A", "B", "C"]
values = [30, 50, 40]

axes[0, 1].bar(categories, values, color='green', label = 'Bar')
axes[0, 1].set_title("Category Comparison")
axes[0, 1].legend()

#Bottom-Left: Distribution

axes[1, 0].hist(sales, bins=10, color='purple', edgecolor='black', label = 'Distribution')
axes[1, 0].set_title("Sales Distribution")
axes[1, 0].legend()


#Bottom-Right: Relationship Analysis

profit = sales * np.random.uniform(0.5, 1.5, 100)

axes[1, 1].scatter(sales, profit, label = "Scatter")
axes[1, 1].set_title("Sales vs Profit Relationship")
axes[1, 1].set_xlabel("Sales")
axes[1, 1].set_ylabel("Profit")
axes[1, 1].legend()

plt.tight_layout()
plt.savefig("Mini Dashboard.png", dpi = 200)
plt.show()
````

###Key Insight: 2D Axes Behavior

###When you use:

``plt.subplots(2, 2)``

The axes object becomes a 2D matrix.

So you access plots like:

- axes[0, 0] → top-left
- axes[0, 1] → top-right
- axes[1, 0] → bottom-left
- axes[1, 1] → bottom-right

##Analyst Interpretation

This layout is used when building complete dashboards, where each subplot answers a different business question:

- Trend → Is performance improving?
- Comparison → Which category is best?
- Distribution → How is data spread?
- Relationship → Are variables connected?

#2. Controlling Layout Structure (VERY IMPORTANT)
## 1. Why Layout Control is a Core Analyst Skill

In real-world dashboards, the biggest problem is not plotting data.

It is:

- overlapping labels
- cramped charts
- unreadable text
- poor visual hierarchy

 A technically correct plot can still look unprofessional if layout is not controlled.

##2. figsize — Dashboard Dimensions
### What is figsize?

It controls the overall size of the figure (dashboard canvas).

``fig, axes = plt.subplots(figsize=(10, 8))``
###Why it matters

If figsize is:

- too small → everything overlaps
- too large → wasted space
- balanced → clean, readable dashboard

````python
fig, axes = plt.subplots(2, 2, figsize=(10, 6))

x = np.linspace(0, 10, 100)
axes[0, 0].plot(x, np.sin(x))
axes[0, 1].plot(x, np.cos(x))
axes[1, 0].plot(x, x**2)
axes[1, 1].plot(x, np.sqrt(x))

plt.show()
````

##3. plt.tight_layout() — Automatic Spacing Fix
### What problem does it solve?

When multiple subplots are used:

- titles overlap
- labels collide
- spacing becomes uneven
###Solution
`plt.tight_layout()`
###What it does

It automatically:

- adjusts spacing between plots
- prevents overlap
- aligns elements neatly

````python
fig, axes = plt.subplots(2, 2, figsize=(8, 5))

x = np.linspace(0, 10, 100)

axes[0, 0].plot(x, np.sin(x))
axes[0, 1].plot(x, np.cos(x))
axes[1, 0].plot(x, x**2)
axes[1, 1].plot(x, np.sqrt(x))

plt.tight_layout()
plt.show()
````

###Analyst Insight

 Always use tight_layout() in dashboards unless you manually control spacing.

##4. Manual Spacing Control (subplots_adjust)
 Why do we need this?

Sometimes tight_layout() is not enough when:

- building complex dashboards
- designing presentation-level visuals
- controlling exact spacing

###Syntax
`plt.subplots_adjust(wspace=0.3, hspace=0.4)`
### Meaning
- wspace → horizontal spacing (left-right gap)
- hspace → vertical spacing (top-bottom gap)

````python
fig, axes = plt.subplots(2, 2, figsize=(8, 5))

x = np.linspace(0, 10, 100)

axes[0, 0].plot(x, np.sin(x))
axes[0, 1].plot(x, np.cos(x))
axes[1, 0].plot(x, x**2)
axes[1, 1].plot(x, np.sqrt(x))

plt.subplots_adjust(wspace=0.5, hspace=0.5)

plt.show()
````

###Key Insight
- tight_layout() → automatic control (safe default)
- subplots_adjust() → manual precision control (advanced)

#3. Designing Multi-Plot Dashboards
##1. What “Dashboard Design” Really Means

A multi-plot dashboard is not just a collection of graphs.

It is a structured analytical story that answers multiple business questions in one view.

Instead of asking:

“What plots should I make?”

You should ask:

“What questions about the data should this dashboard answer?”

##2. Core Analytical Structure of a Good Dashboard

A professional dashboard usually follows this logic:

- Trend (Line Plot) → What is happening over time?
- Comparison (Bar Chart) → Which category performs best?
- Distribution (Histogram) → How is the data spread?
- Relationship (Scatter Plot) → Are variables connected?

##3. Building a Realistic Business Dataset

We simulate a simple business scenario:

Sales data over time
Product categories
Profit behavior

###Step 1: Data Setup

````python
np.random.seed(42)

days = np.arange(1, 101)

sales = np.random.randint(20, 100, 100)

profit = sales * np.random.uniform(0.5, 1.5, 100)

categories = ["Electronics", "Clothing", "Home", "Sports"]
category_sales = [320, 250, 400, 180]
````

##4. Designing the Dashboard (2x2 Layout)
### Important Principle

Each subplot must answer ONE question.



````python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

#Trend Analysis (Line Plot)
#What question does it answer?
#How are sales changing over time?
axes[0, 0].plot(days, sales, color='blue')
axes[0, 0].set_title("Sales Trend Over Time")
axes[0, 0].set_xlabel("Days")
axes[0, 0].set_ylabel("Sales")
axes[0, 0].grid(True)

#Category Comparison (Bar Chart)
#What question does it answer?
#Which category performs best?
axes[0, 1].bar(categories, category_sales, color='green')
axes[0, 1].set_title("Category Sales Comparison")
axes[0, 1].set_xlabel("Category")
axes[0, 1].set_ylabel("Sales")

#Distribution Analysis (Histogram)
#What question does it answer?
#How are sales values distributed?
axes[1, 0].hist(sales, bins=10, color='purple', edgecolor='black')
axes[1, 0].set_title("Sales Distribution")
axes[1, 0].set_xlabel("Sales Range")
axes[1, 0].set_ylabel("Frequency")

#Relationship Analysis (Scatter Plot)
#What question does it answer?
#Is there a relationship between sales and profit?

axes[1, 1].scatter(sales, profit, color='red')
axes[1, 1].set_title("Sales vs Profit Relationship")
axes[1, 1].set_xlabel("Sales")
axes[1, 1].set_ylabel("Profit")
axes[1, 1].grid(True)

plt.tight_layout()
plt.show()
````

###Insight Line Chart
- Shows fluctuations in sales
- Helps identify peaks and drops
- Useful for demand forecasting

###Insight Bar Chart
- Electronics and Home dominate sales
- Sports is underperforming
- Helps in business decision-making (marketing focus)

###Insight Distribution Analysis
- Shows whether sales are consistent or highly variable
- Helps detect outliers
- Useful for risk analysis

###Insight Scatter Plot
- Higher sales generally lead to higher profit
- But not perfectly linear (real-world behavior)
- Shows variability in profit margins

##4. Common Issue (IMPORTANT)

If you use:

It may overlap or cut the main title

````python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

x = np.arange(1, 101)
sales = np.random.randint(20, 100, 100)

axes[0, 0].plot(x, sales)
axes[0, 0].set_title("Sales Trend")

axes[0, 1].bar(["A", "B", "C"], [30, 50, 40])
axes[0, 1].set_title("Category Comparison")

axes[1, 0].hist(sales, bins=10)
axes[1, 0].set_title("Distribution")

profit = sales * np.random.uniform(0.5, 1.5, 100)
axes[1, 1].scatter(sales, profit)
axes[1, 1].set_title("Sales vs Profit")

fig.suptitle("Business Sales Dashboard (Q1 Analysis)", fontsize=16)

plt.tight_layout(rect = [0, 0, 1, 0.95])
plt.show()
````

``plt.tight_layout(rect = [0, 0, 1, 0.95])``
It defines the area of the figure where subplots are allowed to fit.

|Value|	Meaning|
|-----|--------|
|0 (left)|	Start from extreme left|
|0 (bottom)|	Start from bottom|
|1 (right)|	Go to extreme right|
|0.95 (top)|	Stop a bit before the top|




###Key Insight
- Values range from 0 to 1
- They represent percentage of figure size

So:

- 0 = 0% (edge)

- 1 = 100% (edge)

##5. Full Styled Dashboard Example

````python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

days = np.arange(1, 101)
sales = np.random.randint(20, 100, 100)
profit = sales * np.random.uniform(0.5, 1.5, 100)

categories = ["Electronics", "Clothing", "Home", "Sports"]
category_sales = [320, 250, 400, 180]

# Trend
axes[0, 0].plot(days, sales, color='blue', label='Sales')
axes[0, 0].set_title("Sales Trend")
axes[0, 0].grid(True)
axes[0, 0].legend()

# Comparison
axes[0, 1].bar(categories, category_sales, color='green', label='Category Sales')
axes[0, 1].set_title("Category Comparison")
axes[0, 1].legend()

# Distribution
axes[1, 0].hist(sales, bins=10, color='purple', edgecolor='black', label='Distribution')
axes[1, 0].set_title("Sales Distribution")
axes[1, 0].legend()

# Relationship
axes[1, 1].scatter(sales, profit, color='red', label='Sales vs Profit')
axes[1, 1].set_title("Relationship")
axes[1, 1].grid(True)
axes[1, 1].legend()

fig.suptitle("Business Dashboard", fontsize=16)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
````

##4. Building an Insight-Oriented Dashboard


````python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

days = np.arange(1, 101)
sales = np.random.randint(20, 100, 100)
profit = sales * np.random.uniform(0.5, 1.5, 100)

categories = ["Electronics", "Clothing", "Home", "Sports"]
category_sales = [320, 250, 400, 180]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

axes[0, 0].plot(days, sales, color='blue')
axes[0, 0].set_title("Sales Trend (Is growth happening?)")
axes[0, 0].set_xlabel("Days")
axes[0, 0].set_ylabel("Sales")
axes[0, 0].grid(True)

axes[0, 1].bar(categories, category_sales, color='green')
axes[0, 1].set_title("Category Performance (Which performs best?)")
axes[0, 1].set_ylabel("Sales")

axes[1, 0].hist(sales, bins=10, color='purple', edgecolor='black')
axes[1, 0].set_title("Sales Distribution (How is data spread?)")
axes[1, 0].set_xlabel("Sales Range")


axes[1, 1].scatter(sales, profit, color='red')
axes[1, 1].set_title("Sales vs Profit (Is there a relationship?)")
axes[1, 1].set_xlabel("Sales")
axes[1, 1].set_ylabel("Profit")
axes[1, 1].grid(True)

fig.suptitle("Business Insight Dashboard", fontsize=16)

plt.tight_layout(rect=[0, 0, 1, 0.95])
plt.show()
````

- overwhelms the viewer  
- reduces clarity  
- hides important insights  

---

##  Example of Bad Thinking

> “I know 10 plot types → I will use all of them”

---

##  Correct Approach

- Use **only necessary plots**
- Each plot must answer a **unique question**

---

##  Rule

> If removing a chart does not reduce insight → remove it

---

---

#  3. Mistake 2: No Clear Story

---

##  The Problem

Plots are placed randomly without logical flow.

 Viewer does not understand:

- where to look first  
- what the main insight is  

---

##  Example of Bad Dashboard

- Trend placed randomly  
- Comparison at bottom  
- No logical reading order  

---

##  Correct Approach

Follow **analytical flow**:

1. Trend (top-left)  
2. Comparison (top-right)  
3. Distribution (bottom-left)  
4. Relationship (bottom-right)  

---

##  Rule

> A dashboard must guide the viewer step-by-step

---

---

#  4. Mistake 3: Inconsistent Styles

---

##  The Problem

- random colors  
- some plots have grid, others don’t  
- different font sizes  
- mismatched styling  

 Result: looks messy and unprofessional

---

##  Example Issues

- Blue line + red bars + yellow histogram (no meaning)  
- Legends missing or placed randomly  

---

##  Correct Approach

- Use **consistent color scheme**
- Apply grid intentionally
- Maintain uniform styling

---

##  Rule

> All plots should look like part of the same system

---

---

#  5. Mistake 4: No Insights

---

##  The Biggest Mistake

Dashboard shows data, but does NOT explain anything.

---

##  Example

- “Sales Trend”
- “Bar Chart”
- “Histogram”

 These are **descriptions**, not insights

---

##  Correct Approach

Titles should reflect **questions or insights**:

- “Sales Trend (Is growth happening?)”
- “Category Performance (Which performs best?)”

---

##  Analyst Thinking

Every plot should answer:

> “So what?”

---

##  Rule

> A chart without insight is just decoration

---

- no structure  
- no consistent styling  
- no clear purpose

##  Professional Dashboard

- 4–5 focused charts  
- logical flow  
- consistent design  
- each chart answers a question
