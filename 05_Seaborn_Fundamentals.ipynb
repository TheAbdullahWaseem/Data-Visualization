# **1. Introduction to Seaborn**

---

## 1. What is Seaborn?

Seaborn is a **high-level data visualization library built on top of Matplotlib**.

It is designed to:

- create **statistically meaningful visualizations**
- work directly with **structured data (Pandas DataFrames)**
- produce **clean, professional plots by default**

---

## Core Idea

> Matplotlib gives you control  
> Seaborn gives you clarity

---

## Why Seaborn was created

Matplotlib is powerful, but:

- requires more code for styling  
- plots look basic by default  
- not optimized for statistical visualization  

Seaborn solves these problems by:

- adding **better default styles**
- simplifying syntax
- focusing on **data analysis use-cases**

---

##  2. Seaborn vs Matplotlib

---

##  Conceptual Difference

| Feature | Matplotlib | Seaborn |
|--------|----------|--------|
| Level | Low-level | High-level |
| Control | Full control | Controlled abstraction |
| Styling | Manual | Automatic |
| Data handling | Arrays/lists | DataFrames |
| Use case | Custom plots | Analytical plots |

---

##  1. Understanding Seaborn Function Structure

Seaborn follows a **standard, consistent function structure** across most plots.

###  General Pattern

```python
sns.plot_type(x=..., y=..., data=...)

###What this means
- x → variable for horizontal axis
- y → variable for vertical axis
- data → dataset (usually a Pandas DataFrame

Seaborn is designed to work with data columns, not just raw arrays

##2. Using x, y, and data


````python
df = pd.DataFrame({
    "Day": [1, 2, 3, 4, 5],
    "Sales": [10, 20, 15, 25, 30]
})

sns.lineplot(x="Day", y="Sales", data=df)

plt.title("Sales Trend")
plt.show()
````

In real-world data analysis:

- data comes in tables (DataFrames)
- not separate lists or arrays

````python
sns.scatterplot(x="Day", y="Sales", data=df)
plt.show()
````

##4. Array-Based vs DataFrame-Based Plotting
###Array-Based (Matplotlib Style)

````python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 20, 15, 25, 30])

plt.plot(x, y)
plt.show()
````

###Key Differences

#  1. Why These Plots Matter

In data analysis, most real-world questions fall into **four core categories**:

- How is something changing over time?
- Which category performs better?
- How is the data distributed?
- Is there a relationship between variables?
- Are there outliers or unusual behavior?

 Seaborn provides **specialized functions** for each of these.

---

##  1. Line Plot — `sns.lineplot()`

---

###  Purpose

Used to analyze:

> **Trend over time**

---

###  When to use

- sales over days/months
- revenue growth
- user growth trends

---

###  What it tells you

- Increasing trend
- Decreasing trend
- Fluctuations
- Seasonality patterns

---

###  Example

```python id="lineplot_example"
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Day": np.arange(1, 11),
    "Sales": [10, 15, 13, 18, 25, 30, 28, 35, 40, 45]
})

sns.lineplot(x="Day", y="Sales", data=df)
plt.title("Sales Trend Over Time")

````python
df = pd.DataFrame({
    "Day": np.arange(1, 11),
    "Sales": [10, 15, 13, 18, 25, 30, 28, 35, 40, 45]
})

sns.lineplot(x = 'Day', y = 'Sales', data = df)
plt.show()
````

##2. Bar Plot — sns.barplot()
### Purpose

###Used for:

Comparing categories
### When to use
- product performance
- category-wise sales
- region comparisons
### What it tells you
- which category is highest
- ranking between groups
- performance gaps
### Example

````python
df = pd.DataFrame({
    "Category": ["A", "B", "C", "D"],
    "Sales": [30, 50, 40, 60]
})

sns.barplot(x="Category", y="Sales", data=df)
plt.title("Category-wise Sales Comparison")
plt.show()
````

##3. Histogram — sns.histplot()
##Purpose

###Used for:

Understanding distribution of data

### When to use
- salary distribution
- sales distribution
- age distribution
### What it tells you
- most common values
- spread of data
skewness
- outliers (rough idea)
### Example

````python
data = np.random.randint(1, 100, 100)

sns.histplot(data, bins = 10, kde=True)
plt.title("Data Distribution")
plt.show()
````

##4. Scatter Plot — sns.scatterplot()
##Purpose

###Used for:

Finding relationship between two variables

### When to use
- sales vs profit
- study hours vs marks
- advertising vs revenue
### What it tells you
- correlation (positive/negative)
- strength of relationship
- clusters in data
### Example

````python
df = pd.DataFrame({
    "Sales": [10, 20, 30, 40, 50],
    "Profit": [5, 12, 18, 25, 30]
})

sns.scatterplot(x="Sales", y="Profit", data=df)
plt.title("Sales vs Profit Relationship")
plt.show()
````

##5. Box Plot — sns.boxplot()
## Purpose

###Used for:

Detecting distribution and outliers

### When to use
- salary analysis
- exam score distribution
- comparing variability between groups
### What it shows
- median (middle value)
- quartiles (Q1, Q3)
- spread (IQR)
- outliers (extreme values)
### Example

````python
df = pd.DataFrame({
    "Department": ["A", "A", "A", "B", "B", "B"],
    "Salary": [50, 55, 60, 70, 75, 120]
})

sns.boxplot(x="Department", y="Salary", data=df)
plt.title("Salary Distribution by Department")
plt.show()
````

##Final Analyst Understanding
##Each plot answers a specific question:


| Plot Type     | Analytical Question                                      |
|--------------|----------------------------------------------------------|
| Line Plot     | Is there a trend over time?                              |
| Bar Plot      | Which category performs best?                            |
| Histogram     | How is the data distributed?                             |
| Scatter Plot  | Is there a relationship between two variables?          |
| Box Plot      | Are there outliers and how is the data spread?          |

# **4.  Built-in Styling System in Seaborn**

##  1. Why Styling Matters in Data Analysis

In real-world analytics, raw plots are not enough.

 Seaborn solves this by providing **built-in styling systems** that remove the need for manual customization.

##  2. `sns.set_style()` — Base Visual Theme

---

###  What it does

It controls the **background and grid appearance** of all plots.

 Think of it as setting the “visual environment” for your dashboard.

---

###  Available Styles

### 1. `whitegrid`
- White background + grid lines
- Best for analysis

### 2. `darkgrid`
- Darker background + grid
- Useful for highlighting contrast

### 3. `white`
- Clean white background
- No grid lines

### 4. `dark`
- Dark background
- More visual emphasis

### 5. `ticks`
- Minimal style
- Adds tick marks for precision

---

###  Example Usage

```python id="set_style_example"
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

sns.set_style("whitegrid")

df = pd.DataFrame({
    "x": np.arange(1, 6),
    "y": [10, 20, 15, 25, 30]
})

sns.lineplot(x="x", y="y", data=df)
plt.title("Using whitegrid Style")
plt.show()

````python
sns.set_style("darkgrid")

df = pd.DataFrame({
    "x": np.arange(1, 6),
    "y": [10, 20, 15, 25, 30]
})

sns.lineplot(x = 'x', y = 'y', data = df)
plt.show()
````

##3. sns.set_theme() — Global Styling Control
### What it does

It sets the overall visual theme for all Seaborn plots:

- font size
- scaling
- context (paper, notebook, talk, poster)
### Example

````python
sns.set_theme(style = 'darkgrid', context = 'notebook')
sns.lineplot(x = 'x', y = 'y', data = df)
plt.title("Using set_theme")
plt.show()
````

##4. Default Color Palettes
### What is a palette?

A color palette is a predefined set of colors used across plots.

### Common Palettes
- deep (default)
- muted
- bright
- pastel
- dark
- colorblind
###Example

````python
sns.set_palette("pastel")

sns.barplot(x=["A", "B", "C"], y=[10, 20, 30])
plt.title("Using Pastel Palette")
plt.show()
````

````python
sns.set_theme(style = 'darkgrid', context = 'notebook', palette = 'muted')

sns.barplot(x=["A", "B", "C"], y=[10, 20, 30])
plt.title("Using Pastel Palette")
plt.show()
````

##5. Why Seaborn Reduces Manual Styling
### Matplotlib Approach

You manually set:

- colors
- grid
- fonts
- styles

 More control, but more effort

### Seaborn Approach

You just set:

##  What it does

##4. Long-Form vs Wide-Form Data (Basic Understanding)
##  Long-Form Data (Preferred in Seaborn)

Each row represents one observation.

| Day | Sales |
|-----|------|
| 1   | 10   |
| 2   | 20   |
| 3   | 15   |

 This is Seaborn’s ideal format

---

##  Wide-Form Data

Each column represents a separate variable.

| Day | Sales_A | Sales_B |
|-----|--------|--------|
| 1   | 10     | 12     |
| 2   | 20     | 18     |

##Key Difference
|Type|	Structure|	Seaborn Preference|
|-----|----------|--------|
|Long-form|	tidy rows|Best|
|Wide-form|	multiple columns|	Limited use|

###Real-world datasets (CSV, Excel, SQL) are almost always long-form

# **7.  Basic Customization in Seaborn**

---

##  1. Why Customization Still Matters

Seaborn gives clean plots by default, but:

- defaults are not always enough  
- business reports require clarity  
- labels and titles must communicate meaning  

 A professional analyst always **refines the plot for communication**

---

##  2. Setting Titles

---

##  Purpose

The title tells the viewer:

> “What is this chart about?”

---

##  Weak Title

- "Line Plot"  
- "Graph 1"  

---

##  Strong Title

- "Sales Trend Over Time"  
- "Monthly Revenue Growth"  

---

##  Code


````python

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

df = pd.DataFrame({
    "Day": [1, 2, 3, 4, 5],
    "Sales": [10, 20, 15, 25, 30]
})

sns.lineplot(x="Day", y="Sales", data=df)

plt.title("Sales Trend Over Time")

plt.show()
````

##3. Axis Labels (xlabel, ylabel)
### Purpose

Axis labels explain:

- what each axis represents
- units of measurement

````python
sns.lineplot(x = 'Day', y = 'Sales', data = df)
plt.title('Sales Trend')
plt.xlabel('Days')
plt.ylabel('Sales Amount')

plt.show()
````

##4. Basic Color Control
### Why control color?
- highlight important data
- maintain consistency
- improve readability
### Example

````python
sns.set_theme(style = 'darkgrid', context = 'notebook', palette = 'muted')

sns.lineplot(x = 'Day', y = 'Sales', data = df, color = 'green')
plt.title("Sales Trend (Green Line)")
plt.show()
````

##5. Grid Behavior (Automatic vs Manual)
###Default Behavior

Seaborn automatically applies grid (depending on style):

````python
sns.set_style('darkgrid')
sns.lineplot(x = 'Day', y = 'Sales', data = df)

plt.title('Auto Grid Example')
plt.show()
````

###Manual Control (Using Matplotlib)

You can still control grid manually:

````python
sns.lineplot(x="Day", y="Sales", data=df)

plt.grid(True)

plt.title("Manual Grid Control")
plt.show()
````

##6. Combining Everything (Professional Example)
###Code

````python
sns.set_style("whitegrid")

sns.lineplot(x="Day", y="Sales", data=df, color="blue")

plt.title("Sales Trend Over Time")
plt.xlabel("Days")
plt.ylabel("Sales")

plt.grid(True)

plt.show()
````
