#2. Two Interfaces of Matplotlib
##2.1 Why Matplotlib Has Two Interfaces?

Matplotlib provides two different ways to create visualizations because it was designed for both:
- Quick exploratory analysis
- Professional-level data visualization systems

These two approaches are:
1. pyplot (State-based interface)
2. Object-Oriented (OO) interface

##2.2 Pyplot (State Based Interface)
pyplot is a **simple, state-based interface** in Matplotlib that allows users to create plots quickly using a MATLAB-like style.

It automatically manages figures and axes in the background.
###Key Characteristics
- Easy to use
- Minimal code required
- Good for quick analysis
- Not suitable for complex or professional projects

```python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y)
plt.title("Simple Plot using pyplot")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")
plt.show()
```

In pyplot:

- We directly call functions like `plt.plot()`
- Matplotlib internally creates Figure and Axes automatically
- We do NOT explicitly manage objects

This makes it fast but less flexible.

##Use pyplot when:
- You are quickly exploring data
- You are testing ideas
- You are doing simple plots

Avoid using it for complex dashboards or production code.

##2.3 Object-Oriented (OO) Interface
###Definition
The Object-Oriented (OO) interface is a **professional way of creating plots** in Matplotlib where we explicitly create and manage:

- Figure (the entire canvas)
- Axes (the actual plotting area)

This gives full control over every part of the visualization.

##Core Components
###1. Figure
A Figure is the **complete canvas or window** where plots are drawn.

Think of it as the “page” of a report.
###2. Axes
Axes are the **actual plotting area inside the figure** where data is visualized.

A Figure can contain one or multiple Axes.


```python
x = [1,2,3,4]
y = [10,20,15,25]

fig, ax = plt.subplots()

ax.plot(x, y)
ax.set_title("Object Oriented Plot")
ax.set_xlabel("X Axis")
ax.set_ylabel("Y Axis")


plt.show()
```

In this approach:

- We explicitly create a Figure and Axes using `plt.subplots()`
- We use `ax.plot()` instead of `plt.plot()`
- We control every element manually

This gives full flexibility and control.

In real-world data analysis and production systems:

1. Complex dashboards are required
2. Multiple plots are combined
3. Fine control over layout is needed
4. Code must be reusable and scalable

##Key Advantages of OO Approach
- Better control over plots
- Easier to manage multiple charts
- More readable in large projects
- Standard in data science teams
- Required for advanced visualization

##2.4 pyplot vs Object-Oriented Comparison
| Feature | pyplot | Object-Oriented |
|----------|--------|----------------|
| Ease of use | Very easy | Moderate |
| Control | Low | High |
| Professional use | Limited | High |
| Scalability | Poor | Excellent |
| Recommended | Learning / quick plots | Industry / production |

##Side-by-Side Code Comparison

```python
plt.plot([1,2,3], [10,20,30])
plt.title("pyplot style")
plt.show()
```

```python
fig, ax = plt.subplots()

ax.plot([1,2,3], [10,20,30])
ax.set_title("OO style")

plt.show()
```

- pyplot is for quick testing and exploration
- Object-Oriented is for real data analysis work
- Figure = canvas
- Axes = plotting area
- Professionals always prefer OO approach

#3. Core Architecture of Matplotlib (Figure, Axes, and Hierarchy)
##3.1 Why Understanding Architecture Matters
Matplotlib is not just a plotting library—it is a structured system.

Most beginners struggle because they skip the internal structure and only memorize functions.

Understanding the architecture helps you:
- Debug plotting issues easily
- Build complex visualizations
- Work professionally with dashboards
- Understand how plots are actually constructed

##3.2 The Matplotlib Hierarchy
Matplotlib follows a hierarchical structure:

1. Figure (Top Level Container)
2. Axes (Plotting Area inside Figure)
3. Axis (X-axis and Y-axis inside Axes)
4. Plot Elements (Lines, Bars, Text, etc.)

Think of it like this:

- Figure = Entire page of a report
- Axes = One chart inside the page
- Axis = X and Y scale of that chart
- Plot elements = actual data lines/bars

##3.3 Figure
A Figure is the **highest level container** in Matplotlib.

It represents the entire canvas where one or more plots are drawn.

``fig = plt.figure()``

``plt.show()``

Here:

- A blank figure is created
- No data is plotted yet
- This is just the canvas

In real projects, figures hold multiple plots.


##3.4 Axes
Axes is the **actual plotting area** where data is visualized.

This is where graphs, charts, and plots are drawn.

```python
fig, ax = plt.subplots()

plt.show()
```

In this code:

- `fig` = entire figure (canvas)
- `ax` = single plot area inside figure
- `plt.subplots()` automatically creates both

This is the most commonly used pattern in professional work.

##3.5 Axis (X and Y Components)
Axis refers to the **X-axis and Y-axis scales** inside a plot.

They define:
- What values are shown
- How data is scaled
- How ticks and labels appear

```python
fig, ax = plt.subplots()

ax.set_xlabel("Months")
ax.set_ylabel("Sales")

plt.show()
```

Here:

- X-axis = Months
- Y-axis = Sales values
- Labels help interpret the graph

Without axis labels, plots are meaningless in real-world analysis.

```python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]

fig, ax = plt.subplots()

ax.plot(months, sales)
ax.set_title("Monthly Sales Trend")
ax.set_xlabel("Months")
ax.set_ylabel("Sales")

plt.show()
```

This example includes the full Matplotlib structure:

- Figure → created by plt.subplots()
- Axes → ax object
- Axis → x and y labels
- Plot → line chart

This is the standard professional structure.

##Multiple Axes Plot
We use multiple axes to:

- Compare different charts side by side
- Create dashboards
- Analyze multiple variables together
- Improve storytelling with data

```python
import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4]
y1 = [10, 20, 15, 25]
y2 = [5, 15, 10, 20]
y3 = [2, 8, 6, 12]
y4 = [7, 14, 21, 28]

# Create figure and axes (2 rows, 2 columns)
fig, ax = plt.subplots(2, 2)

# Plot in each subplot
ax[0, 0].plot(x, y1)
ax[0, 0].set_title("Plot 1")

ax[0, 1].plot(x, y2)
ax[0, 1].set_title("Plot 2")

ax[1, 0].plot(x, y3)
ax[1, 0].set_title("Plot 3")

ax[1, 1].plot(x, y4)
ax[1, 1].set_title("Plot 4")

# Adjust layout to avoid overlapping
plt.tight_layout()

plt.show()
```

##Cleaner Version

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 2)

ax[0, 0].plot([1, 2, 3], [10, 20, 30])
ax[0, 1].plot([1, 2, 3], [30, 20, 10])
ax[1, 0].plot([1, 2, 3], [5, 15, 25])
ax[1, 1].plot([1, 2, 3], [25, 15, 5])

for row in ax:
    for a in row:
        a.set_title("Subplot")

plt.tight_layout()
plt.show()
```

#4. Basic Line Plot
##4.1 What is a Line Plot?
A line plot is a type of visualization used to display data points connected by straight lines.

It is mainly used to show how data changes over time or in a sequence.

##4.2 Why Line Plots are Important in Data Analysis
Line plots are one of the most important tools in data analysis because they help in:

##4.4 Creating a Simple Line Plot

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

plt.plot(x, y)
plt.show()
```

##4.5 Line Plot with Real-World Context

##4.6 Understanding “Trend”
###What is a Trend?
A trend is the general direction in which data is moving over time.

It can be:
- Upward trend (growth)
- Downward trend (decline)
- Flat trend (no change)

```python
x = [1, 2, 3, 4, 5]

upward = [10, 20, 30, 40, 50]
downward = [50, 40, 30, 20, 10]
flat = [20, 20, 20, 20, 20]

fig, ax = plt.subplots(1,3)

ax[0].plot(x, upward)
ax[0].set_title("Upward Trend")

ax[1].plot(x, downward)
ax[1].set_title("Downward Trend")

ax[2].plot(x, flat)
ax[2].set_title("Flat Trend")

plt.tight_layout()
plt.show()
```

This visualization shows three types of trends:

- Upward → growth (positive business performance)
- Downward → decline (problem signal)
- Flat → no change (stagnation)

Understanding trends is one of the most important skills for a data analyst.

##4.7 Professional Line Plot (Proper Formatting)

```python
months = ["Jan", "Feb", "Mar", "Apr", "May"]
sales = [100, 120, 90, 150, 200]

plt.plot(months, sales, marker='o')

plt.title("Monthly Sales Trend")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```

This version is professional because:

#6. Legends
## 6.1 What is a Legend?

A legend is a visual guide that explains what each line, color, or marker in a plot represents.

It maps plotted data (lines or points) to their corresponding labels.

In simple terms, a legend answers:
"Which line represents which data?"

```python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]
profit = [30, 50, 40, 70]

plt.plot(months, sales)
plt.plot(months, profit)

plt.title("Sales and Profit Trend")
plt.xlabel("Months")
plt.ylabel("Amount")

plt.show()
```

In this plot:

- Two lines are shown
- But we don't know which line represents sales or profit

This makes the visualization confusing and unusable.

## 6.2 Why Legends are Important

Legends are essential because they:

1. Identify each data series clearly
2. Prevent confusion when multiple lines exist
3. Improve readability of charts
4. Make visualizations self-explanatory

Without a legend, a multi-line chart loses its meaning.

## 6.3 How to Add a Legend (Basic Method)

To create a legend, each plot must have a label.

Then we call the legend function to display it.

```python
plt.plot(months, sales, label = 'Sales')
plt.plot(months, profit, label='Profit')

plt.title("Sales and Profit Trend")
plt.xlabel("Months")
plt.ylabel("Amount")

plt.legend()

plt.show()
```

- label="Sales" assigns a name to the first line
- label="Profit" assigns a name to the second line
- plt.legend() displays the legend box

Now the viewer can clearly identify each line.

## 6.4 Legend in Object-Oriented Approach (Professional Way)

In professional code, legends are added using the Axes object.

```python
fig, ax = plt.subplots()

ax.plot(months, sales, label = "Sales")
ax.plot(months, profit, label = 'Profit')

ax.set_title("Sales and Profit Trend")
ax.set_xlabel("Months")
ax.set_ylabel("Amount")

ax.legend()

plt.plot()
```

- Keeps code structured and scalable
- Necessary when working with multiple subplots
- Preferred in real-world projects

## 6.5 Legend Positioning

Legends can be placed in different locations to avoid overlapping with data.

Common positions:
- upper right
- upper left
- lower right
- lower left

```python
plt.plot(months, sales, label = 'Sales')
plt.plot(months, sales, label = 'Profit')

plt.legend(loc = 'upper left')
plt.show()
```

```python
plt.plot(months, sales, label = 'Sales')
plt.plot(months, sales, label = 'Profit')

plt.legend(loc = 'lower right')
plt.show()
```

### 6.6 Legend Styling

Legends can be customized to improve appearance.

You can control:
- Font size
- Title
- Frame visibility

```python
fig, ax = plt.subplots()

ax.plot(months, sales, label = 'Sales', marker = 'o')
ax.plot(months, profit, label = 'Amount', marker = '*')

ax.set_title("Months and Sales Trend", fontsize = 12, fontweight='bold')
ax.set_title("Sales and Profit Trend")

ax.legend(loc = 'lower right', title = 'Metrics', fontsize =10)
plt.show()
```

## 6.7 Common Mistakes

Beginners often:

- Forget to add labels → legend doesn't work
- Call legend() without labels → empty legend
- Place legend over important data
- Use unclear names (e.g., "Line1", "Data")

A legend must always be meaningful and clear.

## 6.8 Analyst Mindset

When working with multiple datasets:

- Always label each data series
- Ensure the legend is readable
- Avoid clutter

Your goal is clarity, not decoration.

## 6.9 What You MUST Remember (Critical)

- Legends explain multiple data series
- Always use label= in plots
- Always call legend() after plotting
- Use Object-Oriented approach (ax.legend()) in professional work
- A multi-line chart without a legend is incomplete

## 7.2 Line Color

Line color controls the color of the plotted data line.

It helps differentiate between multiple datasets and improves visual clarity.

```python
x = [1, 2, 3, 4]
y = [10, 20, 15, 25]

plt.plot(x, y, color = 'blue')

plt.title("Line Color Example")
plt.show()
```

## 7.3 Line Style

Line style controls the pattern of the line.

It helps distinguish between different datasets when colors alone are not enough.

```python
plt.plot(x, y, color = 'blue', linestyle = '--', label = 'Sales')

plt.title("Line Style Example")

plt.show()
```

Common line styles:

- "-"  → solid line
- "--" → dashed line
- "-." → dash-dot line
- ":"  → dotted line

```python
plt.plot(x, y, linestyle=":")

plt.title("Line Style Example")
plt.show()
```

### 7.4 Markers

Markers highlight individual data points on a plot.

They help in:
- Identifying exact data positions
- Improving visibility
- Making trends clearer

```python
plt.plot(x, y, marker="o")

plt.title("Marker Example")
plt.show()
```

Common markers:

- "o" → circle
- "s" → square
- "^" → triangle
- "*" → star
- "x" → cross

```python
plt.plot(x, y, marker="^")

plt.title("Marker Example")
plt.show()
```

## 7.5 Combining Styling Elements

In real-world analysis, we combine color, linestyle, and markers to create clear and distinguishable plots.

```python
plt.plot(x, y, label='Sales', color= 'red',linestyle= '-.', marker='*')

plt.title("Combined Styling")
plt.xlabel("X Axis")
plt.ylabel("Y Axis")

plt.show()
```

Combining styles helps:

- Make plots visually clear
- Differentiate multiple lines
- Improve presentation quality

This is expected in professional work.

##7.6 Styling Multiple Lines

```python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]
profit = [30, 50, 40, 70]

plt.plot(months, sales, label = 'Sales', linestyle='--',color='blue', marker='^')
plt.plot(months, profit, label='Amount',linestyle='--', color='red', marker='s')

plt.title("Sales vs Profit", fontsize='10', fontweight='bold')
plt.xlabel("Months")
plt.ylabel("Amount")

plt.legend(loc='upper right', title='Metrics', fontsize=10)

plt.show()
```

#Saving Plots
## 8.1 Why Saving Plots is Important

In real-world data analysis, plots are not just displayed—they are saved and shared.

Saving plots is important because:

- Plots are included in reports and presentations
- They are uploaded to portfolios (GitHub, LinkedIn)
- They are shared with stakeholders
- They are used in dashboards and documentation

A data analyst must be able to export visualizations properly.

## 8.2 What is savefig()?

`savefig()` is a function in Matplotlib used to save a plot as an image file.

It allows you to export your visualization into different file formats such as PNG, JPG, or PDF.

```python
months = ["Jan", "Feb", "Mar", "Apr"]
sales = [100, 150, 120, 180]
profit = [30, 50, 40, 70]

plt.plot(months, sales, label = 'Sales', linestyle='--',color='blue', marker='^')
plt.plot(months, profit, label='Amount',linestyle='--', color='red', marker='s')

plt.title("Sales vs Profit", fontsize='10', fontweight='bold')
plt.xlabel("Months")
plt.ylabel("Amount")

plt.legend(loc='upper right', title='Metrics', fontsize=10)

plt.savefig("Sales_plot.png")
plt.show()
```

- "sales_plot.png" is the file name
- The plot is saved in the current working directory
- The format is determined by the file extension (.png)

Important:
Always call savefig() BEFORE plt.show()

## 8.4 File Formats

Matplotlib supports multiple file formats. The most commonly used are:

1. PNG
   - High quality
   - Lossless format
   - Best for reports and portfolios

2. JPG (or JPEG)
   - Smaller file size
   - Slight loss in quality
   - Useful for quick sharing

3. PDF
   - Vector format (scalable)
   - Best for professional reports and printing
   
``plt.savefig("plot.png")``
   Recommended for portfolio
``plt.savefig("plot.jpg")``    Smaller size
``plt.savefig("plot.pdf")``    Professional reports

## 8.5 Saving with High Quality

Quality of saved plots is controlled using the `dpi` parameter (dots per inch).

Higher DPI means better resolution.

```python
plt.plot(months, sales)

plt.title("Monthly Sales Trend")

plt.savefig("high_quality_plot.png", dpi=300)

plt.show()
```

- dpi=300 is standard for high-quality images
- Recommended for:
  - Portfolio
  - Reports
  - Presentations

## 8.6 Saving Without Extra White Space

Sometimes saved plots contain unnecessary white space.

This can be fixed using:
- bbox_inches='tight'

```python
plt.plot(months, sales)

plt.title("Monthly Sales Trend")

plt.savefig("clean_plot.png", bbox_inches='tight')

plt.show()
```

## 8.7 Saving Using Object-Oriented Approach

In professional code, plots are saved using the Figure object.

```python
fig, ax = plt.subplots()

ax.plot(months, sales)
ax.set_title("Monthly Sales Trend")

fig.savefig('Oo_plot.png', dpi= 300, bbox_inches='tight')
plt.show()
```

- Keeps code structured
- Useful when working with multiple plots
- Preferred in real-world projects
