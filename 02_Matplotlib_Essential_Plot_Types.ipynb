#Line Plot
## 2.1 What is a Line Plot?
A line plot is a type of visualization that connects data points using straight lines to show how a variable changes over a continuous dimension (usually time).

It is mainly used to observe **trends and patterns over time**.

A line plot answers one main question:

**"How is something changing over time?"**

## 2.2 When to Use a Line Plot?
A line plot should be used when:

###  1. Data is time-based
- Days, months, years, hours

###  2. You want to see change or trend
- Increase or decrease over time

###  3. You are analyzing continuous data
- Sales, profit, temperature, stock prices

###Do NOT use line plots when:

- Data is categorical (e.g., countries, products)
- You are only comparing groups (use bar chart instead)

## 2.3 Time Series Interpretation
A time series means data that is ordered by time.

Example:

Month → Sales

Jan → 100

Feb → 120

Mar → 90

1. Is the trend increasing or decreasing?
2. Are there sudden drops or spikes?
3. Is the pattern stable or volatile?
4. What happened at peaks and dips?

###Business Thinking Example

If Sales:

- Increase → business growing
- Decrease → possible issue
- Fluctuating → unstable demand

## 2.4 How to Extract Insights from a Line Plot
A data analyst does NOT just draw a line plot.

###They extract meaning from it.

####Step 1: Observe direction
- Upward / downward / flat

####Step 2: Identify changes
- sudden increase or drop

####Step 3: Explain possible reason
- marketing campaign
- seasonality
- product issue

####Step 4: Conclude business insight

###Example Insight

If sales go:

Jan → 100

Feb → 150

Mar → 90

Apr → 200

####Insight:

- Sales are unstable
- March shows a drop (possible issue)
- April shows recovery (possible improvement or campaign)

## 2.5 Multiple Lines Comparison
A line plot can also show multiple variables to compare trends over time.

```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May"]

sales = [100, 120, 90, 150, 200]
profit = [20, 30, 10, 50, 70]

plt.plot(months, sales, label="Sales", marker='o')
plt.plot(months, profit, label="Profit", marker='o')

plt.title("Sales vs Profit Trend")
plt.xlabel("Months")
plt.ylabel("Value")

plt.legend()

plt.show()
```

####When comparing multiple lines:

1. Which one is growing faster?
2. Is profit following sales or not?
3. Is there a gap between variables?

Example insight:
If sales increase but profit does not → cost issue

## 2.6 Common Mistakes
Beginners often:

- Just plot without interpretation
- Ignore time order
- Use line plots for categorical data
- Forget to compare trends meaningfully

## 2.7 Final Analyst Mindset
A professional data analyst does NOT think:

**"I made a line plot"**

They think:

- "What story does this trend tell?"
- "What changed over time?"
- "What action should business take?"

## 2.8 What You MUST Remember
- Line plot = trend over time
- Best for continuous/time-based data
- Always ask “why is it changing?”
- Multiple lines = comparison of trends
- Insight matters more than graph

#3. Bar Chart
## 3.1 What is a Bar Chart?
A bar chart is a type of visualization that uses rectangular bars to represent data values.

The length (or height) of each bar is proportional to the value it represents.

It is mainly used to compare values across different categories.

```python
categories = ["A", "B", "C", "D"]
values = [10, 25, 15, 30]

plt.bar(categories, values)

plt.show()
```

- Each category is shown on the X-axis
- Values are shown as bar height
- Taller bar = higher value

This is the simplest form of comparison visualization.

## 3.2 When to Use a Bar Chart?
A bar chart is used when:

- You want to compare different categories
- Data is categorical (not continuous)
- You want to identify highest and lowest values
- You need ranking or comparison

```python
products = ["Laptop", "Phone", "Tablet", "Monitor"]
sales = [500, 800, 300, 450]

plt.bar(products, sales)

plt.title("Product Sales Comparison")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

From this chart:

- Phone has highest sales
- Tablet has lowest sales
- We can clearly compare product performance

This is how businesses identify best-selling products.

## 3.3 Vertical Bar Chart
A vertical bar chart displays bars upright (default orientation).

It is best used when category names are short and readable.

```python
plt.bar(["Jan", "Feb", "Mar", "Apr"], [100, 150, 90, 200])

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```

- X-axis shows categories (months)
- Y-axis shows values (sales)
- Each bar represents one month

This is the most commonly used bar chart type.

## 3.4 Horizontal Bar Chart
A horizontal bar chart displays bars sideways.

It is useful when category names are long or when you want better readability.

```python
plt.barh(products, sales)

plt.title("Product Sales Comparison")
plt.xlabel("Sales")
plt.ylabel("Products")

plt.show()
```

- Bars go left to right instead of bottom to top
- Easier to read long category names
- Often used in ranking dashboards

## 3.5 Category Comparison
Bar charts are widely used in business to compare:

- Product performance
- Department performance
- Country-wise sales
- Monthly revenue comparison

```python
products = ["Laptop", "Phone", "Tablet", "Monitor", "Headphones"]
sales = [500, 800, 300, 450, 600]

plt.bar(products, sales, color="green")

plt.title("Sales by Product")
plt.xlabel("Products")
plt.ylabel("Sales")

plt.show()
```

From this chart:

- Phone is top performer
- Tablet is weakest product
- Headphones show strong demand

Business decisions:
- Increase marketing for Tablets
- Invest more in Phones and Headphones

## 3.6 Common Mistakes (VERY IMPORTANT)
Beginners often:

- Use bar charts for time series (wrong → use line plot)
- Do not label axes
- Use too many categories (cluttered chart)
- Ignore interpretation (just plotting without insight)

## 3.7 What You MUST Remember
- Bar charts compare categories
- Best for discrete data (not time)
- Vertical = default, Horizontal = readability
- Always interpret results, not just plot
- Business value comes from comparison insight

#4. Histogram (Distribution Analysis)
##4.1 What is a Histogram
A histogram is a type of visualization that shows the distribution of numerical data by grouping values into intervals called "bins".

It helps us understand how data is spread across different value ranges.

###A histogram answers one key question:

**"How are my values distributed?"**

###Simple Example Concept

Imagine exam scores:

0–20 → 2 students

20–40 → 5 students

40–60 → 10 students

60–80 → 8 students

80–100 → 3 students

This grouping is a histogram.

## 4.2 Histogram vs Bar Chart (VERY IMPORTANT)
Although histograms and bar charts look similar, they are completely different.

###Key Differences
| Feature | Histogram | Bar Chart |
|----------|-----------|-----------|
| Data type | Numerical (continuous) | Categorical |
| Purpose | Distribution | Comparison |
| Bars | Touch each other | Separated |
| X-axis | Ranges (bins) | Categories |

- Bar chart → compares things (Laptop vs Phone)
- Histogram → shows distribution (how many values fall in ranges)

## 4.3 Frequency Distribution
A frequency distribution shows how often values occur within specific ranges.

###Example Concept
Scores:

10, 15, 20, 22, 25, 30, 35, 40, 45, 50

###Grouped:

0–20 → 2 values  

20–40 → 5 values  

40–60 → 3 values

```python
data = [10, 15, 20, 22, 25, 30, 35, 40, 45, 50]

plt.hist(data)

plt.title("Basic Histogram")
plt.xlabel("Value Range")
plt.ylabel("Frequency")

plt.show()
```

## 4.4 Bins Concept
Bins are the intervals used to group data in a histogram.

Each bin represents a range of values.
###Bins control:

- How detailed your distribution is
- How smooth or rough your histogram looks
- How easy it is to interpret data

###Example Concept
If data is exam scores:

Bins = 5

0–20

20–40

40–60

60–80

80–100

```python
data = [10, 15, 20, 22, 25, 30, 35, 40, 45, 50]

plt.hist(data, bins=5)

plt.title("Histogram with Bins")
plt.xlabel("Value Range")
plt.ylabel("Frequency")

plt.show()
```

```python
import numpy as np

data = np.random.normal(50, 15, 200)

plt.hist(data, bins=10, edgecolor= 'black')

plt.title("Distribution of Data")
plt.xlabel("Values")
plt.ylabel("Frequency")

plt.show()
```

## 4.5 How to Interpret a Histogram (ANALYST SKILL)
A histogram helps you understand:

### 1. Shape of Data
- Normal (bell-shaped)
- Skewed left
- Skewed right

### 2. Spread
- Narrow = consistent data
- Wide = varied data

### 3. Peaks
- High frequency areas

If salaries are histogrammed:

- Most people in 30k–50k → middle-class distribution
- Few high salaries → right skewed

This helps HR and business planning.

## 4.6 Common Mistakes
Beginners often:

- Use histogram for categorical data (wrong)
- Ignore bins selection
- Misinterpret shape of distribution
- Confuse histogram with bar chart

## 4.7 What You MUST Remember
- Histogram = distribution of numerical data
- Bars touch each other (continuous data)
- Bins control grouping of data
- Used to understand shape, spread, and frequency
- Essential for statistics and machine learning

#. Scatter Plot
## 5.1 What is a Scatter Plot?
A scatter plot is a visualization that shows the relationship between two numerical variables by plotting data points on a 2D plane.

Each point represents one observation with:
- X-axis → one variable
- Y-axis → another variable

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 5]

plt.scatter(x, y)

plt.title("Basic Scatter Plot")
plt.xlabel("X Values")
plt.ylabel("Y Values")

plt.show()
```

A scatter plot answers one key question:

 "Is there a relationship between two variables?"

## 5.2 Relationship Between Two Variables
In a scatter plot, each point represents a pair of values.

Example:
- X → Advertising Spend
- Y → Sales

Each dot = one data record

```python
ads = [10, 20, 30, 40, 50]
sales = [100, 150, 200, 250, 300]

plt.scatter(ads, sales)

plt.title("Ads vs Sales")
plt.xlabel("Advertising Spend")
plt.ylabel("Sales")

plt.show()
```

As advertising increases, sales also increase.

 This suggests a relationship between advertising and sales.
 ## 5.3 Correlation Intuition
Correlation describes how two variables move in relation to each other.

###1. Positive Correlation
As X increases, Y also increases.

Example:
- More ads → more sales

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]

plt.scatter(x, y)
plt.title("Positive Correlation")

plt.show()
```

###2. Negative Correlation
As X increases, Y decreases.

Example:
- Higher price → lower demand

```python
x = [1, 2, 3, 4, 5]
y = [50, 40, 30, 20, 10]

plt.scatter(x, y)
plt.title("Negative Correlation")

plt.show()
```

###3. No Correlation
There is no clear relationship between variables.

Example:
- Shoe size vs intelligence

```python
x = [1, 2, 3, 4, 5]
y = [23, 45, 12, 67, 34]

plt.scatter(x, y)
plt.title("No Correlation")

plt.show()
```

## 5.4 Real-World Use Cases
Scatter plots are widely used in:

- Marketing → Ads vs Sales
- Health → Height vs Weight
- Finance → Risk vs Return
- Education → Study hours vs marks

```python
study_hours = [1, 2, 3, 4, 5, 6]
marks = [30, 40, 50, 60, 70, 80]

plt.scatter(study_hours, marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

From this plot:

- More study hours → higher marks
- Strong positive relationship

Business/real-world conclusion:
👉 Increasing effort improves performance

## 5.5 How to Extract Insights
When analyzing a scatter plot, ask:

1. Is there a pattern?
2. Is it positive or negative?
3. How strong is the relationship?
4. Are there any unusual points (outliers)?

## 5.6 Common Mistakes
Beginners often:

- Assume correlation = causation (WRONG)
- Ignore outliers
- Use scatter for categorical data
- Do not interpret patterns

## 5.7 What You MUST Remember
- Scatter plot = relationship between two variables
- Each point = one observation
- Positive, negative, or no correlation
- Used heavily in data science and machine learning
- Insight matters more than plotting

# 6. Box Plot (Outlier Detection)
## 6.1 What is a Box Plot?
A box plot (also called a box-and-whisker plot) is a visualization used to summarize the distribution of numerical data using five key statistics:

- Minimum
- First Quartile (Q1)
- Median (Q2)
- Third Quartile (Q3)
- Maximum

It is mainly used to understand data spread and detect outliers.

A box plot answers:

- "How is my data spread and are there any unusual values?"

```python
import matplotlib.pyplot as plt

data = [10, 20, 30, 40, 50, 60, 70]

plt.boxplot(data)

plt.title("Basic Box Plot")

plt.show()
```

## 6.2 Five Number Summary (Foundation)
A box plot is based on five important values:

1. Minimum → smallest value
2. Q1 (First Quartile)
3. Median (Q2)
4. Q3 (Third Quartile)
5. Maximum → largest value

## 6.3 What are Quartiles?
Quartiles divide your sorted data into four equal parts.

Q1, Q2, Q3 are the points that split the data.

Take this data:

```python
data = [10, 20, 30, 40, 50, 60, 70]
```

###Step 1: Sort Data
``10, 20, 30, 40, 50, 60, 70``

###Step 2: Find Median (Q2)
``Median = middle value = 40``

 This divides data into two halves

###Step 3: Find Q1
Left half = [10, 20, 30]

Q1 = median of left half = 20

###Step 4: Find Q3
Right half = [50, 60, 70]

Q3 = median of right half = 60

###Final Values

Minimum = 10  

Q1 = 20  

Median = 40  

Q3 = 60  

Maximum = 70  

## 6.4 Structure of Box Plot
A box plot consists of:

- Box → from Q1 to Q3
- Line inside box → Median
- Whiskers → extend to min and max (or near them)

###Visual Meaning
- Box = middle 50% of data
- Median line = center of data
- Whiskers = spread of data

## 6.5 Outliers
Outliers are values that are significantly different from the rest of the data.

They can indicate:
- Errors
- Rare events
- Important anomalies

###How Outliers Are Detected
Outliers are detected using IQR (Interquartile Range):

IQR = Q3 - Q1

Lower bound = Q1 - 1.5 × IQR  
Upper bound = Q3 + 1.5 × IQR

###Example
Data:
[10, 20, 30, 40, 50, 60, 100]

Q1 = 20  
Q3 = 60  
IQR = 40  

Upper bound = 60 + 60 = 120  

 100 is within range → not extreme  
But if value was 200 → outlier

```python
data = [10, 20, 30, 40, 50, 60, 200]

plt.boxplot(data)

plt.title("Box Plot with Outlier")

plt.show()
```

- Most data lies in box
- 200 appears far away → outlier

 This signals something unusual in data

## 6.6 Why Analysts Use Box Plots
Box plots are used in:

### 1. Data Cleaning
- Detect wrong values
- Identify anomalies

### 2. Data Understanding
- See spread of data
- Understand variability

### 3. Feature Analysis
- Important in machine learning

###Real Example

If salaries dataset shows:

Most salaries → 30k–50k  
Few salaries → 200k  

 Those high values may be:
- Executives
- Data errors

Analyst must investigate

## 6.7 Common Mistakes
- Thinking boxplot shows exact values (it shows summary)
- Not understanding quartiles
- Ignoring outliers
- Confusing whiskers with min/max always

## 6.8 What You MUST Remember
- Box plot = distribution summary
- Quartiles divide data into 4 parts
- Median = center of data
- IQR = spread of middle data
- Outliers = unusual values
- Used heavily in data cleaning

#Understanding Box Plot Terms
## 1. Q1, Q2, Q3 (Quartiles)

Quartiles divide sorted data into 4 equal parts.

- Q1 → First Quartile (25% of data is below this)
- Q2 → Median (50% of data is below this)
- Q3 → Third Quartile (75% of data is below this)

###Example
data = [10, 20, 30, 40, 50, 60, 70]


Q1 = 20  

Q2 (Median) = 40  

Q3 = 60  

##2. The Box

The box in a boxplot represents the middle 50% of data.

- Start of box → Q1  
- End of box → Q3
###Meaning
- Box starts at Q1
- Box ends at Q3
- Line inside box = Median (Q2)

##3. Interquartile Range (IQR)
IQR = Q3 - Q1

It measures the spread of the middle 50% of data.

###Example
Q1 = 20  

Q3 = 60  

IQR = 60 - 20 = 40

##4. Lower Bound & Upper Bound
These are limits used to detect outliers.

Lower Bound = Q1 - 1.5 × IQR  

Upper Bound = Q3 + 1.5 × IQR

###Example
Q1 = 20  

Q3 = 60  

IQR = 40  

Lower Bound = 20 - (1.5 × 40) = -40  
Upper Bound = 60 + (1.5 × 40) = 120
###Meaning
Values below lower bound → outliers

Values above upper bound → outliers

##5. Whiskers
Whiskers are lines extending from the box to the smallest and largest values within the bounds.

###Important Detail

Whiskers do NOT always go to min/max.

They go to:

Smallest value ≥ lower bound

Largest value ≤ upper bound
###Example
data = [10, 20, 30, 40, 50, 60, 200]

Upper Bound = 120  

Whisker stops at 60  

200 is outside → shown as outlier

##6. Outliers
Outliers are values outside the lower and upper bounds.

###Example
Value = 200  
Upper Bound = 120  

 200 is an outlier

SUPER SIMPLE SUMMARY
- Q1 → where box starts
- Q3 → where box ends
- Median → line inside box
- IQR → size of box (Q3 - Q1)
- Whiskers → lines extending from box
- Bounds → define outliers
- Outliers → separate points

##Difference between IQR and Median
- Median → line inside the box
- IQR → length of the box (Q3 - Q1)
- Median = where the center is
- IQR = how wide the middle data is

## 1. Median (Q2)
The median is the middle value of a sorted dataset.

It divides the data into two equal halves:
- 50% of values are below it
- 50% of values are above it
The median tells you:

 "What is the center of the data?"

 data = [10, 20, 30, 40, 50]

Median = 30

## 2. IQR (Interquartile Range)
IQR = Q3 - Q1

It measures the spread of the middle 50% of the data.
The IQR tells you:

 "How spread out is the central portion of the data?"

 data = [10, 20, 30, 40, 50]

Q1 = 20  

Q3 = 40  

IQR = 40 - 20 = 20
